```python
import numpy as np
import matplotlib.pyplot as plt

prelu_history = np.load("/content/VAEAC_Thesis/celeba_model/prelu_history.npy", allow_pickle=True)
max_len = max(len(x) for x in prelu_history)
history_padded = np.array([np.pad(x, (0, max_len - len(x)), constant_values=np.nan) for x in prelu_history])

plt.figure(figsize=(12,5))
for i in range(max_len):
    plt.plot(history_padded[:, i], alpha=0.5)
plt.xlabel('Epoch')
plt.ylabel('PReLU Negative Slope (alpha)')
plt.title('Evolution of PReLU Parameters During Training')
plt.tight_layout()
plt.show()
```

```python
import torch
import numpy as np
from importlib import import_module
from os.path import join

# ==== CONFIG ====
MODEL_DIR = "celeba_model"      # <-- change to your model dir if needed
CHECKPOINT = "best_checkpoint.tar"   # or 'last_checkpoint.tar'
USE_CUDA = torch.cuda.is_available()

# ==== LOAD MODEL ====
model_module = import_module(f"{MODEL_DIR}.model")
from VAEAC import VAEAC

model = VAEAC(
    model_module.reconstruction_log_prob,
    model_module.proposal_network,
    model_module.prior_network,
    model_module.generative_network
)
if USE_CUDA:
    model = model.cuda()

checkpoint = torch.load(join(MODEL_DIR, CHECKPOINT), map_location='cuda' if USE_CUDA else 'cpu')
model.load_state_dict(checkpoint['model_state_dict'])

# ==== EXTRACT PReLU ALPHAS ====
alphas = []
for name, module in model.named_modules():
    if isinstance(module, torch.nn.PReLU):
        value = module.weight.data.cpu().numpy().flatten()
        alphas.extend(value)
        print(f"{name}: {value}")

alphas = np.array(alphas)
print("\nAll PReLU alphas:", alphas)
print("Mean alpha:", alphas.mean())
print("Median alpha:", np.median(alphas))
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import os

result_dir = "celeba_inpaintings"
n_show = 6  # how many examples to display

for i in range(n_show):
    fig, axs = plt.subplots(1, 5, figsize=(18, 4))
    files = [
        f"{i:05d}_groundtruth.jpg",
        f"{i:05d}_input.jpg",
        f"{i:05d}_sample_000.jpg",
        f"{i:05d}_sample_001.jpg",
        f"{i:05d}_sample_002.jpg",
        f"{i:05d}_sample_003.jpg",
        f"{i:05d}_sample_004.jpg",
        f"{i:05d}_sample_005.jpg"
    ]
    titles = ["Groundtruth", "Masked input", "Inpainted 1", "Inpainted 2", "Inpainted 3"]
    for j, (fname, title) in enumerate(zip(files, titles)):
        img = Image.open(os.path.join(result_dir, fname))
        axs[j].imshow(img)
        axs[j].set_title(title)
        axs[j].axis("off")
    plt.tight_layout()
    plt.show()
```
